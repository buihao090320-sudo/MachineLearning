Bài 1. Tạo Tensor cơ bản

Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.

```python
import torch
a = torch.tensor(
    [[1, 2, 3],
    [4, 5, 6]],
    dtype=torch.int32
)
print(a)
print("Kích thước:", a.shape)
print("Số chiều:", a.ndim)
print("Kiểu dữ liệu:", a.dtype)
```

Bài 2. Tạo các Tensor thường dùng

Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

```python
import torch
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)
print(a)
print(b)
print(c)
print(d)
print(e)
f = torch.full((5, 5), 10)
print(f)
```

Bài 3. Kiểm tra thông tin Tensor

Sử dụng shape, size(), ndim và dtype để kiểm tra các thuộc tính của Tensor.

```python
import torch

a = torch.zeros(2, 3, 4)
print("Tensor: ")
print(a)
print("Shape", a.shape)
print("Size", a.size())
print("Number of dimesions:", a.ndim)
print("Data type:", a.dtype)
```
