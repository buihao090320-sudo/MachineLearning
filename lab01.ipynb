Bài 5. Reshape và Flatten
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

```python
import torch
x = torch.tensor(
    3.6,
    requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
y = 3 * x**2 + 2 * x + 1
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

    if epoch % 100 == 0:
        print(f"Epoch {epoch}: MSE = {mse.item():.6f}")

print("\nHệ số học được:\n", w)
print("\nHệ số thật:\n [1, 2, 3]")
```

Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000

optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("Nghiệm tìm được:")
print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

print("\nKiểm tra:")
print("A + B =", (A + B).item(), "(cần bằng 9)")
print("C - D =", (C - D).item(), "(cần bằng 1)")
print("A + C =", (A + C).item(), "(cần bằng 8)")
print("B - D =", (B - D).item(), "(cần bằng 2)")

print("\nSai số:")
print("y1 =", (A + B - 9).item())
print("y2 =", (C - D - 1).item())
print("y3 =", (A + C - 8).item())
print("y4 =", (B - D - 2).item())

print("\nKết luận:")
if (
    abs((A + B - 9).item()) < 1e-4 and
    abs((C - D - 1).item()) < 1e-4 and
    abs((A + C - 8).item()) < 1e-4 and
    abs((B - D - 2).item()) < 1e-4
):
    print("4 phương trình đều được thỏa mãn.")
else:
    print("Nghiệm chưa hội tụ chính xác.")
```

• Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
print("Tính giá trị hàm số:")
print("x =", x.item())
print("y =", y.item())
y.backward()
print("\nAutograd:")
print("dy/dx =", x.grad.item())
dao_ham_thu_cong = 4 * 2 + 5
print("\nTính thủ công:")
print("y' = 4x + 5")
print("y'(2) =", dao_ham_thu_cong)
print("\nSo sánh:")
if x.grad.item() == dao_ham_thu_cong:
    print("✓ Autograd và tính thủ công giống nhau!")
else:
    print("✗ Hai kết quả khác nhau!")

x_plot = np.linspace(-5, 5, 200)
y_plot = 2 * x_plot**2 + 5 * x_plot + 3
x0 = 2
y0 = 2 * x0**2 + 5 * x0 + 3
slope = 13
tangent = slope * (x_plot - x0) + y0
plt.figure(figsize=(9, 6))
plt.plot(x_plot, y_plot, label="y = 2x² + 5x + 3")
plt.plot(x_plot, tangent, "--", label="Tiếp tuyến tại x = 2")
plt.scatter([x0], [y0], s=100, label="Điểm (2, 21)")
plt.axvline(x0, linestyle=":")
plt.axhline(y0, linestyle=":")
plt.annotate(
    "x = 2, y = 21\nĐạo hàm = 13",
    xy=(x0, y0),
    xytext=(2.5, 30),
    arrowprops=dict(arrowstyle="->"),
)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Autograd - Đạo hàm của y = 2x² + 5x + 3")
plt.grid(True)
plt.legend()
plt.show()
```

• Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch
polynomial = np.poly1d([2, -3, 5])
print("Hàm thật:")
print("Y = 2X² - 3X + 5")
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
print("\nSố lượng dữ liệu:", N)
XX = np.hstack([X * X, X, np.ones_like(X)])
print("\nXX có 3 cột:")
print("Cột 1 = X²")
print("Cột 2 = X")
print("Cột 3 = 1")
w = torch.randn(3, 1, requires_grad=True)
print("\nHệ số ban đầu:")
print(w)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
EPOCHS = 2000
for epoch in range(EPOCHS):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    if epoch % 200 == 0:
        print(f"Epoch {epoch}: MSE = {mse.item():.6f}")
print("\nHệ số tìm được:")
print(w.detach().numpy().flatten())
print("\nHệ số thật:")
print([2, -3, 5])
```
