BÀI 1: LÀM QUEN VỚI PYTORCH

Bài 1. Tạo Tensor cơ bản

```python
import torch
a = torch.tensor(
    [[1, 2, 3],
    [4, 5, 6]],
    dtype=torch.int32
)
print(a)
print("Kích thước:", a.shape)
print("Số chiều:", a.ndim)
print("Kiểu dữ liệu:", a.dtype)
```

Bài 2. Tạo các Tensor thường dùng

```python
import torch
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)
print(a)
print(b)
print(c)
print(d)
print(e)
f = torch.full((5, 5), 10)
print(f)
```
